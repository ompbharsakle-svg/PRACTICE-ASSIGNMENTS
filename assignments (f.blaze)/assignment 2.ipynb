# 1. Write a program to check if a number entered by the user is positive, negative, or zero.

```python
#taking input from user
no=eval(input("enter any number ==>"));
if no==0:
    print("zero")
elif no<0:
    print("negative")
else:
    print("positive")
    
```

# 2. Write a Python program to check whether a given number is even or odd. 

```python
#taking input from user
no=int(input("enter any integer"));
#no. is even or not using if statement
if no%2==0:
    print("number is even")
else:
    print("number is odd")
```

# 3 .Take a year as input from the user and check whether it is a leap year or not. 

```python
#taking input from user
year=int(input("enter year"));
if year%4==0:
    print("it is a leap year");
else:
    print("it isn't a leap year");
```

# 5 .Write a program to print even numbers from 1 to 50 using a while loop. 

```python
#initializing i with 1
i=1;
#while() loop 
while(i<=50):
    if i%2==0:
        print(i);
    i+=1;
```

# 6. Write a program that asks the user to enter numbers continuously and stops when the user enters 0. (Use break statement)

```python
#initializing the while() loop
while True:
    no = int(input("Enter a number (0 to stop): "));
#breaks the statement when the num is 0
    if no == 0:
        break;

    print("You entered:", no);

print("Program stopped.");
```

# 7. Print numbers from 1 to 10, but skip printing 5. (Use continue statement) 

```python
# using for() loop
for no in range(1,11):
    if no==5:
        continue;
    print(no);
```

# 9. Write a program to take 3 numbers as input and print the largest number among them.

```python
# taking inpute from user
a=eval(input("enter value of a"));
b=eval(input("enter value of b"));
c=eval(input("enter value of c"));
#using elif ladder
if a>b and a>c:
    print(a,"is greatest among three");
elif b>a and b>c:
    print(b,"is greatest among three");
else:
    print(c,"is greatest among three")
```

# 10. Write a program to create a simple menu-based calculator using if-elif (Addition,Subtraction, Multiplication, Division). 


```python
# initializing two variables
a=int(input("enter 1st number"));
ope=input("");
b=int(input("enter 2st number"));
#for addition
if ope== "+":
    print(a+b);
#for subtraction
elif ope=="-":
    print(a-b);
#for multiplication
elif ope== "*":
    print(a*b);
#for division
elif ope== "/":
    print(a/b);
    
```
