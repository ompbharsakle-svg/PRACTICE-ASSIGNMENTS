9. Write a program to find the intersection and union of two sets.
Set A = {1, 2, 3}, Set B = {3, 4, 5}


```python
A = {1, 2, 3}
B = {3, 4, 5}
# .intersection()
print("intersection = ",A.intersection(B))
# .union()
print("union = ",A.union(B))

```

10. Write a program to store 5 student marks in a list and calculate the average.

```python
#initializing list
marks = []
for i in range(5):
    mark = eval(input("Enter student marks: "))
    marks.append(mark)
#finding average
average = sum(marks) / len(marks)
#display===>print()
print("Student marks:", marks)
print("Average marks:", average)
```
