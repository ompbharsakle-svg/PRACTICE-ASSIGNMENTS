4. Write a function to calculate the factorial of a given number using a loop inside a
function.

```python
no=int(input("enter number to find its factorial"));
def factorial(no):
    res=1
    for i in range(1,no+1):
        res=res*(i);
    print(f"factorial of {no}={res}");
factorial(no)
```

6. Write a function that checks whether a number is prime or not.


```python
# asking user for input
no=int(input(""));
def prime(no):
    count=0;
    for i in range(2,no+1):
        if no%i==0:
            count+=1;
    if count==1:
        print(no,"is prime");
    else:
        print(no,"is not prime");
prime(no);
        

```

7. Create a function that takes a string and returns the number of vowels in that string.

```python
#taking input from user as string
word=input("enter any word");
#defining a function 
def nov(word):
    count=0;
    vowels=["a","e","i","o","u"];
#comparing each element of word with vowels
    for i in range(0,len(word)):
        for j in range(0,len(vowels)):
            if word[i]==vowels[j]:
                count+=1
    print(word);
    print(count);
#calling nov() function
nov(word);
            
```
