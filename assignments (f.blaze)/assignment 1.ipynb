# 5. Write a program to swap two numbers using a third variable.

```python
#initializing 2 variables 
a=36;
b=45;
#value before swaping
print("before=\ta=",a,"\tb=",b)
#taking "temp" variable for swaping
temp=a;
a=b;
b=temp;
#value after swaping
print("after=\ta=",a,"\tb=",b)
```

# 10. Write a program to take marks of 5 subjects from the user and calculate total and percentage.

```python
#taking input from user
py=eval(input("python="));
java=eval(input("java="));
dbms=eval(input("dbms"));
sql=eval(input("sql"));
daa=eval(input("daa"));

#calculation part
total=py+java+dbms+sql+daa;
percent=(total/500)*100;

#output
print(f"total marks = {total} \npercentage = {percent}");
```
